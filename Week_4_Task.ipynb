```python
# ============================================================
# UNIVERSITY STUDENT PERFORMANCE PREDICTION
# END-TO-END MACHINE LEARNING PROJECT
# ============================================================

# =========================
# 1. INSTALL / IMPORT
# =========================

# If libraries are missing in Jupyter/Colab, uncomment:
# !pip install pandas numpy matplotlib seaborn scikit-learn joblib

import warnings
warnings.filterwarnings("ignore")

import os
import zipfile
import urllib.request
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import (
    train_test_split,
    RandomizedSearchCV,
    StratifiedKFold
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler,
    label_binarize
)

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import (
    RandomForestClassifier,
    HistGradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve,
    auc
)

RANDOM_STATE = 42

print("=" * 70)
print("LIBRARIES IMPORTED SUCCESSFULLY")
print("=" * 70)


# =========================
# 2. DOWNLOAD DATASET
# =========================

DATA_URL = (
    "https://archive.ics.uci.edu/static/public/697/"
    "predict+students+dropout+and+academic+success.zip"
)

ZIP_FILE = "student_dropout_dataset.zip"
EXTRACT_FOLDER = "student_dropout_data"

if not os.path.exists(EXTRACT_FOLDER):
    os.makedirs(EXTRACT_FOLDER)

if not os.path.exists(ZIP_FILE):
    print("\nDownloading UCI dataset...")
    urllib.request.urlretrieve(DATA_URL, ZIP_FILE)
    print("Download complete.")
else:
    print("\nDataset ZIP already exists.")

with zipfile.ZipFile(ZIP_FILE, "r") as zip_ref:
    zip_ref.extractall(EXTRACT_FOLDER)

csv_file = None

for root, dirs, files in os.walk(EXTRACT_FOLDER):
    for file in files:
        if file.lower().endswith(".csv"):
            csv_file = os.path.join(root, file)
            break
    if csv_file:
        break

if csv_file is None:
    raise FileNotFoundError("CSV dataset could not be found.")

print("Dataset file:", csv_file)


# =========================
# 3. LOAD DATA
# =========================

df = pd.read_csv(csv_file, sep=";")

print("\n" + "=" * 70)
print("DATASET LOADED")
print("=" * 70)

print("Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())


# =========================
# 4. BASIC EXPLORATION
# =========================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")

for column in df.columns:
    print("-", column)


# =========================
# 5. DATA TYPES + MISSING
# =========================

print("\n" + "=" * 70)
print("DATA TYPES / MISSING VALUES / UNIQUE VALUES")
print("=" * 70)

summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Values": df.isnull().sum().values,
    "Unique_Values": df.nunique().values
})

display(summary)


# =========================
# 6. TARGET DISTRIBUTION
# =========================

TARGET = "Target"

if TARGET not in df.columns:
    raise ValueError(
        f"Target '{TARGET}' not found. "
        f"Available columns: {list(df.columns)}"
    )

print("\n" + "=" * 70)
print("TARGET DISTRIBUTION")
print("=" * 70)

print(df[TARGET].value_counts())

print("\nTarget percentages:")
print(
    (df[TARGET].value_counts(normalize=True) * 100).round(2)
)


# =========================
# 7. TARGET VISUALIZATION
# =========================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x=TARGET,
    order=df[TARGET].value_counts().index
)

plt.title("Distribution of University Student Outcomes")
plt.xlabel("Student Outcome")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()


# =========================
# 8. DUPLICATE CHECK
# =========================

print("\n" + "=" * 70)
print("DUPLICATE CHECK")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicates removed.")
else:
    print("No duplicates found.")


# =========================
# 9. DATA CLEANING
# =========================

print("\n" + "=" * 70)
print("DATA CLEANING")
print("=" * 70)

# Replace infinite values with NaN
df = df.replace([np.inf, -np.inf], np.nan)

# Convert object columns to string type
object_columns = df.select_dtypes(
    include=["object"]
).columns

for col in object_columns:
    df[col] = df[col].astype("string")

print("Infinite values handled.")
print("Object columns converted to string-compatible type.")

print("\nMissing values after cleaning preparation:")

missing_values = df.isnull().sum()

if missing_values.sum() == 0:
    print("No missing values found.")
else:
    display(
        missing_values[missing_values > 0]
        .to_frame("Missing_Values")
    )


# =========================
# 10. SEPARATE X AND y
# =========================

X = df.drop(columns=[TARGET]).copy()
y = df[TARGET].copy()

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget classes:")
print(sorted(y.unique()))


# =========================
# 11. FEATURE ENGINEERING
# =========================

print("\n" + "=" * 70)
print("FEATURE ENGINEERING")
print("=" * 70)


def safe_divide(numerator, denominator):
    """
    Safely divides two pandas Series.
    Returns 0 when denominator is zero.
    """
    denominator = denominator.replace(0, np.nan)

    result = numerator / denominator

    result = result.replace(
        [np.inf, -np.inf],
        np.nan
    )

    return result.fillna(0)


required_columns = [
    "Curricular units 1st sem (enrolled)",
    "Curricular units 2nd sem (enrolled)",
    "Curricular units 1st sem (approved)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 1st sem (evaluations)",
    "Curricular units 2nd sem (evaluations)",
    "Curricular units 1st sem (grade)",
    "Curricular units 2nd sem (grade)"
]

missing_required = [
    col for col in required_columns
    if col not in X.columns
]

if missing_required:
    raise ValueError(
        "Required columns are missing:\n"
        + "\n".join(missing_required)
    )


# Feature 1: Total enrolled units
X["Total_Enrolled_Units"] = (
    X["Curricular units 1st sem (enrolled)"]
    + X["Curricular units 2nd sem (enrolled)"]
)


# Feature 2: Total approved units
X["Total_Approved_Units"] = (
    X["Curricular units 1st sem (approved)"]
    + X["Curricular units 2nd sem (approved)"]
)


# Feature 3: Total evaluated units
X["Total_Evaluated_Units"] = (
    X["Curricular units 1st sem (evaluations)"]
    + X["Curricular units 2nd sem (evaluations)"]
)


# Feature 4: Overall approval rate
X["Overall_Approval_Rate"] = safe_divide(
    X["Total_Approved_Units"],
    X["Total_Enrolled_Units"]
)


# Feature 5: Overall evaluation rate
X["Overall_Evaluation_Rate"] = safe_divide(
    X["Total_Evaluated_Units"],
    X["Total_Enrolled_Units"]
)


# Feature 6: Grade change
X["Grade_Change"] = (
    X["Curricular units 2nd sem (grade)"]
    - X["Curricular units 1st sem (grade)"]
)


# Feature 7: Average grade
X["Average_Grade"] = (
    X["Curricular units 1st sem (grade)"]
    + X["Curricular units 2nd sem (grade)"]
) / 2


engineered_features = [
    "Total_Enrolled_Units",
    "Total_Approved_Units",
    "Total_Evaluated_Units",
    "Overall_Approval_Rate",
    "Overall_Evaluation_Rate",
    "Grade_Change",
    "Average_Grade"
]

print("Created engineered features:")

for feature in engineered_features:
    print("-", feature)

print("\nEngineered feature preview:")

display(
    X[engineered_features].head()
)


# =========================
# 12. FEATURE STATISTICS
# =========================

print("\n" + "=" * 70)
print("ENGINEERED FEATURE STATISTICS")
print("=" * 70)

display(
    X[engineered_features].describe().T
)


# =========================
# 13. OUTLIER HANDLING
# =========================

print("\n" + "=" * 70)
print("OUTLIER HANDLING")
print("=" * 70)


def cap_outliers_iqr(dataframe, columns):

    data = dataframe.copy()

    for col in columns:

        q1 = data[col].quantile(0.25)
        q3 = data[col].quantile(0.75)

        iqr = q3 - q1

        if pd.isna(iqr) or iqr == 0:
            continue

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        data[col] = data[col].clip(
            lower=lower_bound,
            upper=upper_bound
        )

    return data


numeric_columns = X.select_dtypes(
    include=np.number
).columns.tolist()

continuous_like_columns = [
    col
    for col in numeric_columns
    if X[col].nunique(dropna=True) > 10
]

X = cap_outliers_iqr(
    X,
    continuous_like_columns
)

print(
    "Outlier capping applied to",
    len(continuous_like_columns),
    "continuous-like variables."
)

print("No observations were deleted because of outliers.")


# =========================
# 14. TRAIN / TEST SPLIT
# =========================

print("\n" + "=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining class distribution:")

display(
    y_train.value_counts(normalize=True).round(3)
)


print("\nTesting class distribution:")

display(
    y_test.value_counts(normalize=True).round(3)
)


# =========================
# 15. IDENTIFY COLUMN TYPES
# =========================

numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

print("\nNumerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))


# =========================
# 16. PREPROCESSING PIPELINE
# =========================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ],
    remainder="drop"
)

print("\nPreprocessing pipeline created.")


# =========================
# 17. MODEL 1
# LOGISTIC REGRESSION
# =========================

print("\n" + "=" * 70)
print("MODEL 1 — LOGISTIC REGRESSION")
print("=" * 70)

logistic_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            LogisticRegression(
                max_iter=3000,
                class_weight="balanced",
                random_state=RANDOM_STATE
            )
        )
    ]
)


logistic_pipeline.fit(
    X_train,
    y_train
)


logistic_pred = logistic_pipeline.predict(
    X_test
)


logistic_proba = logistic_pipeline.predict_proba(
    X_test
)


logistic_accuracy = accuracy_score(
    y_test,
    logistic_pred
)


logistic_f1 = f1_score(
    y_test,
    logistic_pred,
    average="weighted"
)


logistic_auc = roc_auc_score(
    y_test,
    logistic_proba,
    multi_class="ovr",
    average="weighted"
)


print("Accuracy:", round(logistic_accuracy, 4))
print("Weighted F1:", round(logistic_f1, 4))
print("Weighted ROC-AUC:", round(logistic_auc, 4))


# =========================
# 18. MODEL 2
# RANDOM FOREST
# =========================

print("\n" + "=" * 70)
print("MODEL 2 — RANDOM FOREST")
print("=" * 70)

random_forest_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            RandomForestClassifier(
                n_estimators=300,
                max_depth=None,
                min_samples_split=2,
                min_samples_leaf=1,
                class_weight="balanced",
                random_state=RANDOM_STATE,
                n_jobs=-1
            )
        )
    ]
)


random_forest_pipeline.fit(
    X_train,
    y_train
)


rf_pred = random_forest_pipeline.predict(
    X_test
)


rf_proba = random_forest_pipeline.predict_proba(
    X_test
)


rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)


rf_f1 = f1_score(
    y_test,
    rf_pred,
    average="weighted"
)


rf_auc = roc_auc_score(
    y_test,
    rf_proba,
    multi_class="ovr",
    average="weighted"
)


print("Accuracy:", round(rf_accuracy, 4))
print("Weighted F1:", round(rf_f1, 4))
print("Weighted ROC-AUC:", round(rf_auc, 4))


# =========================
# 19. MODEL 3
# HISTGRADIENTBOOSTING
# =========================

print("\n" + "=" * 70)
print("MODEL 3 — HISTGRADIENTBOOSTING")
print("=" * 70)

gradient_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            SimpleImputer(strategy="median"),
            numeric_features
        ),
        (
            "cat",
            Pipeline(
                steps=[
                    (
                        "imputer",
                        SimpleImputer(
                            strategy="most_frequent"
                        )
                    ),
                    (
                        "onehot",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=False
                        )
                    )
                ]
            ),
            categorical_features
        )
    ],
    remainder="drop"
)


gradient_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            gradient_preprocessor
        ),
        (
            "model",
            HistGradientBoostingClassifier(
                max_iter=300,
                learning_rate=0.08,
                max_leaf_nodes=31,
                random_state=RANDOM_STATE
            )
        )
    ]
)


gradient_pipeline.fit(
    X_train,
    y_train
)


gradient_pred = gradient_pipeline.predict(
    X_test
)


gradient_proba = gradient_pipeline.predict_proba(
    X_test
)


gradient_accuracy = accuracy_score(
    y_test,
    gradient_pred
)


gradient_f1 = f1_score(
    y_test,
    gradient_pred,
    average="weighted"
)


gradient_auc = roc_auc_score(
    y_test,
    gradient_proba,
    multi_class="ovr",
    average="weighted"
)


print("Accuracy:", round(gradient_accuracy, 4))
print("Weighted F1:", round(gradient_f1, 4))
print("Weighted ROC-AUC:", round(gradient_auc, 4))


# =========================
# 20. MODEL COMPARISON
# =========================

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "HistGradientBoosting"
    ],

    "Accuracy": [
        logistic_accuracy,
        rf_accuracy,
        gradient_accuracy
    ],

    "Weighted F1": [
        logistic_f1,
        rf_f1,
        gradient_f1
    ],

    "Weighted ROC-AUC": [
        logistic_auc,
        rf_auc,
        gradient_auc
    ]
})


results = results.sort_values(
    by="Weighted F1",
    ascending=False
).reset_index(drop=True)


display(
    results.round(4)
)


# =========================
# 21. MODEL COMPARISON GRAPH
# =========================

plt.figure(figsize=(10, 6))

results_plot = results.set_index("Model")

results_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Model Performance Comparison")
plt.xlabel("Model")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()


# =========================
# 22. BEST MODEL
# =========================

best_model_name = results.iloc[0]["Model"]

print("\nBest model based on Weighted F1:")
print(best_model_name)


# =========================
# 23. RANDOM FOREST TUNING
# =========================

print("\n" + "=" * 70)
print("HYPERPARAMETER TUNING — RANDOM FOREST")
print("=" * 70)


rf_tuning_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            RandomForestClassifier(
                class_weight="balanced",
                random_state=RANDOM_STATE,
                n_jobs=-1
            )
        )
    ]
)


param_distributions = {

    "model__n_estimators": [
        200,
        300,
        500,
        700
    ],

    "model__max_depth": [
        None,
        10,
        15,
        20,
        30
    ],

    "model__min_samples_split": [
        2,
        5,
        10
    ],

    "model__min_samples_leaf": [
        1,
        2,
        4
    ],

    "model__max_features": [
        "sqrt",
        "log2",
        None
    ]
}


cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)


random_search = RandomizedSearchCV(
    estimator=rf_tuning_pipeline,
    param_distributions=param_distributions,
    n_iter=25,
    scoring="f1_weighted",
    cv=cv,
    verbose=1,
    random_state=RANDOM_STATE,
    n_jobs=-1
)


random_search.fit(
    X_train,
    y_train
)


print("\nBest parameters:")

for parameter, value in random_search.best_params_.items():
    print(parameter, ":", value)


print(
    "\nBest cross-validation Weighted F1:",
    round(random_search.best_score_, 4)
)


# =========================
# 24. FINAL TUNED MODEL
# =========================

best_model = random_search.best_estimator_


# =========================
# 25. FINAL TEST EVALUATION
# =========================

print("\n" + "=" * 70)
print("FINAL TEST SET EVALUATION")
print("=" * 70)


tuned_pred = best_model.predict(
    X_test
)


tuned_proba = best_model.predict_proba(
    X_test
)


tuned_accuracy = accuracy_score(
    y_test,
    tuned_pred
)


tuned_f1 = f1_score(
    y_test,
    tuned_pred,
    average="weighted"
)


tuned_auc = roc_auc_score(
    y_test,
    tuned_proba,
    multi_class="ovr",
    average="weighted"
)


print(
    "Final Accuracy:",
    round(tuned_accuracy, 4)
)

print(
    "Final Weighted F1:",
    round(tuned_f1, 4)
)

print(
    "Final Weighted ROC-AUC:",
    round(tuned_auc, 4)
)


# =========================
# 26. CLASSIFICATION REPORT
# =========================

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        tuned_pred,
        digits=4
    )
)


# =========================
# 27. CONFUSION MATRIX
# =========================

class_names = sorted(
    y.unique()
)


cm = confusion_matrix(
    y_test,
    tuned_pred,
    labels=class_names
)


plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "Confusion Matrix — Tuned Random Forest"
)

plt.xlabel(
    "Predicted Outcome"
)

plt.ylabel(
    "Actual Outcome"
)

plt.tight_layout()
plt.show()


# =========================
# 28. MULTICLASS ROC CURVE
# =========================

y_test_binary = label_binarize(
    y_test,
    classes=class_names
)


plt.figure(figsize=(9, 7))


for i, class_name in enumerate(class_names):

    fpr, tpr, _ = roc_curve(
        y_test_binary[:, i],
        tuned_proba[:, i]
    )

    class_auc = auc(
        fpr,
        tpr
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{class_name} (AUC = {class_auc:.3f})"
    )


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random classifier"
)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "Multiclass ROC Curve — Tuned Random Forest"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# =========================
# 29. FEATURE IMPORTANCE
# =========================

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE")
print("=" * 70)


fitted_preprocessor = (
    best_model.named_steps["preprocessor"]
)


feature_names = (
    fitted_preprocessor
    .get_feature_names_out()
)


rf_model = (
    best_model.named_steps["model"]
)


importance_values = (
    rf_model.feature_importances_
)


feature_importance = pd.DataFrame({

    "Feature": feature_names,

    "Importance": importance_values

})


feature_importance = (
    feature_importance
    .sort_values(
        by="Importance",
        ascending=False
    )
    .reset_index(drop=True)
)


print("Top 20 predictive features:")

display(
    feature_importance.head(20)
)


# =========================
# 30. FEATURE IMPORTANCE GRAPH
# =========================

top_features = (
    feature_importance
    .head(20)
    .sort_values(
        by="Importance"
    )
)


plt.figure(figsize=(10, 8))


plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)


plt.title(
    "Top 20 Features Driving Student Outcome Prediction"
)

plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.tight_layout()
plt.show()


# =========================
# 31. TOP 10 FEATURES
# =========================

print("\n" + "=" * 70)
print("TOP 10 FEATURES")
print("=" * 70)


for i, row in feature_importance.head(10).iterrows():

    print(
        f"{i + 1}. {row['Feature']} "
        f"-> {row['Importance']:.4f}"
    )


# =========================
# 32. KEY FINDINGS
# =========================

top_5 = (
    feature_importance
    .head(5)["Feature"]
    .tolist()
)


print("\n" + "=" * 70)
print("KEY FINDINGS")
print("=" * 70)


print(
    f"""
The tuned Random Forest achieved an accuracy of
{tuned_accuracy:.2%} and a weighted F1 score of
{tuned_f1:.2%} on the held-out test set.

The most influential variables were primarily related
to students' academic engagement and performance,
including course enrollment, evaluations, approvals,
and grades.

The engineered features such as overall approval rate,
evaluation rate, total approved units, and grade change
provide useful summaries of academic progress.

The five highest-ranked model features were:
{", ".join(top_5)}.

Overall, the results suggest that recent academic
engagement and successful completion of curricular units
are important signals for distinguishing students who
drop out, remain enrolled, or graduate.
"""
)


# =========================
# 33. FINAL METRICS TABLE
# =========================

print("\n" + "=" * 70)
print("FINAL MODEL METRICS")
print("=" * 70)


final_results = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Weighted F1",
        "Weighted ROC-AUC"
    ],

    "Score": [
        tuned_accuracy,
        tuned_f1,
        tuned_auc
    ]
})


display(
    final_results.style.format({
        "Score": "{:.4f}"
    })
)


# =========================
# 34. SAVE MODEL
# =========================

MODEL_FILE = (
    "university_student_performance_model.joblib"
)


joblib.dump(
    best_model,
    MODEL_FILE
)


print(
    f"\nModel saved successfully as: {MODEL_FILE}"
)


# =========================
# 35. TEST SAVED MODEL
# =========================

print("\n" + "=" * 70)
print("TESTING SAVED MODEL")
print("=" * 70)


loaded_model = joblib.load(
    MODEL_FILE
)


loaded_predictions = (
    loaded_model.predict(
        X_test.head(10)
    )
)


print(
    "Predictions from loaded model:"
)

print(
    loaded_predictions
)


# =========================
# 36. REUSABLE PREDICTION FUNCTION
# =========================

def predict_student_outcome(
    student_data,
    trained_model
):

    """
    Predict student academic outcome.

    Parameters
    ----------
    student_data : pandas DataFrame
        Student feature data.

    trained_model : fitted sklearn Pipeline
        Trained machine-learning model.

    Returns
    -------
    pandas DataFrame
        Prediction and probabilities.
    """

    predictions = (
        trained_model.predict(
            student_data
        )
    )


    probabilities = (
        trained_model.predict_proba(
            student_data
        )
    )


    output = pd.DataFrame({
        "Predicted_Outcome": predictions
    })


    for i, class_name in enumerate(
        trained_model.classes_
    ):

        output[
            f"Probability_{class_name}"
        ] = probabilities[:, i]


    return output


# Example prediction
prediction_result = predict_student_outcome(
    X_test.head(5),
    loaded_model
)


print("\nExample predictions:")

display(
    prediction_result
)


# =========================
# 37. SAVE RESULTS
# =========================

results.to_csv(
    "model_comparison_results.csv",
    index=False
)


feature_importance.to_csv(
    "feature_importance.csv",
    index=False
)


final_results.to_csv(
    "final_model_metrics.csv",
    index=False
)


print("\n" + "=" * 70)
print("FILES SAVED")
print("=" * 70)

print("""
1. university_student_performance_model.joblib
2. model_comparison_results.csv
3. feature_importance.csv
4. final_model_metrics.csv
""")


# =========================
# 38. FINAL PROJECT SUMMARY
# =========================

print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    f"""
Dataset shape: {df.shape}

Number of original features:
{df.shape[1] - 1}

Number of engineered features:
{len(engineered_features)}

Training samples:
{len(X_train)}

Testing samples:
{len(X_test)}

Models compared:
- Logistic Regression
- Random Forest
- HistGradientBoosting

Final model:
Tuned Random Forest

Final Accuracy:
{tuned_accuracy:.4f}

Final Weighted F1:
{tuned_f1:.4f}

Final Weighted ROC-AUC:
{tuned_auc:.4f}
"""
)

print("=" * 70)
print("END OF PROJECT")
print("=" * 70)
```
